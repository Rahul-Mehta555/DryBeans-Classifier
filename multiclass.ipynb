## 3. Missing Values & Outlier Treatment

```python
print(df.isnull().sum())
```

```python
outlier_rows = []
iqr_bounds = {}

for col in df_numeric.columns:
    q1 = df_numeric[col].quantile(0.25)
    q3 = df_numeric[col].quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    iqr_bounds[col] = (lower_bound, upper_bound)
    count = ((df[col] < lower_bound) | (df[col] > upper_bound)).sum()

    outlier_rows.append({
        'Feature': col,
        'Outlier Count': count,
        'Outlier Percent': round(count / len(df) * 100, 2)
    })

outlier_summary = pd.DataFrame(outlier_rows).sort_values(
    'Outlier Count', ascending=False
)
display(outlier_summary)

# No need to treat outliers as they are not significantly affecting the model performance.

```

## 5. Model Building: Try Multiple Classifiers
Train and test the dataset on a variety of supervised classification algorithms: <br/>
•	Logistic Regression <br/>
•	Decision Tree Classifier <br/>
•	Random Forest Classifier <br/>
•	K-Nearest Neighbors (KNN) <br/>
•	Support Vector Machine (SVM) <br/>
•	Ensemble Learning Methods <br/>
•	Naive Bayes and more… <br/>
Use Cross validation techniques to check if model performance is improved. 


```python
from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import LabelEncoder, PowerTransformer
from sklearn.metrics import classification_report, ConfusionMatrixDisplay
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, ExtraTreesClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB
from xgboost import XGBClassifier

# Inspect skewness; Yeo-Johnson transformation is applied in pipelines
# for models that benefit from transformed, scaled features.
X = df.drop(columns='Class')

labelencoder = LabelEncoder()
y = labelencoder.fit_transform(df['Class'])

skewness = X.skew().sort_values(key=lambda values: values.abs(), ascending=False)
display(skewness.to_frame('Skewness'))

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

transform = PowerTransformer(method='yeo-johnson')

models = {
    'Logistic Regression': make_pipeline(
        transform, LogisticRegression(max_iter=2000, class_weight='balanced')
    ),
    'Decision Tree': DecisionTreeClassifier(
        class_weight='balanced', random_state=42
    ),
    'Random Forest': RandomForestClassifier(
        n_estimators=300, class_weight='balanced_subsample',
        random_state=42, n_jobs=-1
    ),
    'KNN': make_pipeline(
        PowerTransformer(method='yeo-johnson'),
        KNeighborsClassifier(n_neighbors=5, weights='distance')
    ),
    'SVM': make_pipeline(
        PowerTransformer(method='yeo-johnson'),
        SVC(class_weight='balanced')
    ),
    'Naive Bayes': make_pipeline(
        PowerTransformer(method='yeo-johnson'), GaussianNB()
    ),
    'XGBoost': make_pipeline(
        PowerTransformer(method='yeo-johnson'),
        XGBClassifier(
            n_estimators=300, scale_pos_weight=imbalance_ratio,
            random_state=42, n_jobs=-1
        )
    ),
}

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
cv_scores = {}

for name, model in models.items():
    scores = cross_val_score(
        model, X_train, y_train, cv=cv, scoring='f1_macro', n_jobs=-1
    )
    cv_scores[name] = scores.mean()

cv_results = pd.Series(cv_scores, name='Mean CV Macro-F1').sort_values(ascending=False)
display(cv_results.to_frame())

best_name = cv_results.index[0]
best_model = models[best_name]
best_model.fit(X_train, y_train)
y_pred = best_model.predict(X_test)

print(f'Best model by CV Macro-F1: {best_name}')
print(classification_report(y_test, y_pred, zero_division=0))

ConfusionMatrixDisplay.from_predictions(
    y_test, y_pred, xticks_rotation=45, cmap='Blues'
)
plt.tight_layout()
plt.show()
```

## 6. Handling Class Imbalance <br/>
•	Apply techniques like: <br/>
o	SMOTE (Synthetic Minority Over-sampling) <br/>
o	Random Oversampling / Undersampling <br/>
o	Class weighting <br/>
•	Evaluate if performance improves on minority classes


```python
from imblearn.over_sampling import SMOTE

# Fit the transformation on the original training data to avoid leakage.
imbalance_transformer = PowerTransformer(method='yeo-johnson')
X_train_scaled = imbalance_transformer.fit_transform(X_train)
X_test_scaled = imbalance_transformer.transform(X_test)

# Baseline SVM
svm_baseline = SVC()
svm_baseline.fit(X_train_scaled, y_train)
pred_baseline = svm_baseline.predict(X_test_scaled)

# SVM with class weighting
svm_weighted = SVC(class_weight='balanced')
svm_weighted.fit(X_train_scaled, y_train)
pred_weighted = svm_weighted.predict(X_test_scaled)

smote = SMOTE(random_state=42)
X_train_smote, y_train_smote = smote.fit_resample(X_train_scaled, y_train)
svm_smote = SVC()
svm_smote.fit(X_train_smote, y_train_smote)
pred_smote = svm_smote.predict(X_test_scaled)


# Randomly oversample training classes to match the largest class.
X_scaled_df = pd.DataFrame(X_train_scaled, index=X_train.index)
y_train_series = pd.Series(y_train, index=X_train.index)
majority_count = y_train_series.value_counts().max()

oversampled_X, oversampled_y = [], []
for class_id in y_train_series.unique():
    class_indices = y_train_series[y_train_series == class_id].index
    sampled_indices = pd.Series(class_indices).sample(
        n=majority_count,
        replace=len(class_indices) < majority_count,
        random_state=42
    ).to_numpy()

    oversampled_X.append(X_scaled_df.loc[sampled_indices])
    oversampled_y.extend([class_id] * majority_count)

X_train_oversampled = pd.concat(oversampled_X)
svm_oversampled = SVC()
svm_oversampled.fit(X_train_oversampled, oversampled_y)
pred_oversampled = svm_oversampled.predict(X_test_scaled)

# Compare recall and F1 for each class, highlighting the minority classes.
class_counts = pd.Series(y_train).value_counts()
minority_ids = class_counts[class_counts < class_counts.max()].index
class_names = labelencoder.classes_

reports = {
    'Baseline SVM': classification_report(
        y_test, pred_baseline, output_dict=True, zero_division=0
    ),
    'Class-weighted SVM': classification_report(
        y_test, pred_weighted, output_dict=True, zero_division=0
    ),
    'Oversampled SVM': classification_report(
        y_test, pred_oversampled, output_dict=True, zero_division=0
    ),
    'SMOTE SVM': classification_report(
        y_test, pred_smote, output_dict=True, zero_division=0
    )
}

comparison = pd.DataFrame({
    name: {
        class_names[int(class_id)]: {
            'Recall': report[str(class_id)]['recall'],
            'F1': report[str(class_id)]['f1-score']
        }
        for class_id in minority_ids
    }
    for name, report in reports.items()
}).T

display(comparison)
```

Evaluation after class imbalance:
1. Monirity class Bombay performance is best in base.
2. Second minority class is Barbuniya. its performance improved with Class weighted SVM
3. Third minority class is Cali, its performance also improved with class weighted SVM


## 7. Model Evaluation & Overfitting Check
Use appropriate classification metrics: <br/>
•	Accuracy <br/>
•	Precision, Recall, and F1-Score (for each class) <br/>
•	Confusion Matrix <br/>

Check for Overfitting: <br/>
Compare training vs test accuracy and performance metrics.


```python
from sklearn.metrics import accuracy_score, precision_recall_fscore_support

evaluation_rows = []
predictions = {}

for model_name, fitted_model in models.items():
    fitted_model.fit(X_train, y_train)
    train_pred = fitted_model.predict(X_train)
    test_pred = fitted_model.predict(X_test)
    predictions[model_name] = test_pred

    train_precision, train_recall, train_f1, _ = precision_recall_fscore_support(
        y_train, train_pred, average='macro', zero_division=0
    )
    test_precision, test_recall, test_f1, _ = precision_recall_fscore_support(
        y_test, test_pred, average='macro', zero_division=0
    )

    evaluation_rows.append({
        'Model': model_name,
        'Train Accuracy': accuracy_score(y_train, train_pred),
        'Test Accuracy': accuracy_score(y_test, test_pred),
        'Train Macro Precision': train_precision,
        'Test Macro Precision': test_precision,
        'Train Macro Recall': train_recall,
        'Test Macro Recall': test_recall,
        'Train Macro F1': train_f1,
        'Test Macro F1': test_f1,
        'Accuracy Gap': accuracy_score(y_train, train_pred) -
                        accuracy_score(y_test, test_pred)
    })

evaluation_results = (
    pd.DataFrame(evaluation_rows)
    .set_index('Model')
    .sort_values('Test Macro F1', ascending=False)
)
display(evaluation_results.round(3))

# Detailed test metrics and confusion matrix for the model selected by CV
selected_model = models[best_name]
selected_pred = predictions[best_name]

print(f'Per-class test metrics — {best_name}')
print(classification_report(
    y_test,
    selected_pred,
    labels=range(len(class_names)),
    target_names=class_names,
    zero_division=0
))

ConfusionMatrixDisplay.from_predictions(
    y_test,
    selected_pred,
    labels=range(len(class_names)),
    display_labels=class_names,
    xticks_rotation=45,
    cmap='Blues'
)
plt.title(f'{best_name}: Test Confusion Matrix')
plt.tight_layout()
plt.show()
```

Evaluation after classification report <br/>
SVM, Logistic Regression and Naive Bayes are not overfitting
XGBoost, Random Forest, KNN, and Decesion tree seems overfitting

## 8. Hyperparameter Tuning
•	Use GridSearchCV or RandomizedSearchCV to optimize parameters for top-performing models <br/>
•	Document the best parameters and performance improvement


```python
from sklearn.model_selection import GridSearchCV, RandomizedSearchCV
from scipy.stats import loguniform

# Grid search over a focused set of SVM parameters
param_grid = {
    'svc__C': [0.1, 1, 10, 100],
    'svc__gamma': ['scale', 0.001, 0.01, 0.1, 1],
    'svc__kernel': ['rbf']
}

grid_search = GridSearchCV(
    estimator=selected_model,
    param_grid=param_grid,
    scoring='f1_macro',
    cv=cv,
    n_jobs=-1,
    refit=True
)
grid_search.fit(X_train, y_train)

# Randomized search over a broader range of C and gamma values
param_distributions = {
    'svc__C': loguniform(1e-2, 1e3),
    'svc__gamma': loguniform(1e-4, 1e1),
    'svc__kernel': ['rbf']
}

random_search = RandomizedSearchCV(
    estimator=selected_model,
    param_distributions=param_distributions,
    n_iter=25,
    scoring='f1_macro',
    cv=cv,
    n_jobs=-1,
    random_state=42,
    refit=True
)
random_search.fit(X_train, y_train)

print('Grid search best CV macro-F1:', grid_search.best_score_)
print('Grid search best parameters:', grid_search.best_params_)
print('Randomized search best CV macro-F1:', random_search.best_score_)
print('Randomized search best parameters:', random_search.best_params_)

# Keep the best search result as selected_model and evaluate on the test set
best_search = max(
    [grid_search, random_search],
    key=lambda search: search.best_score_
)
selected_model = best_search.best_estimator_
selected_pred = selected_model.predict(X_test)

print(f'\nSelected search: {type(best_search).__name__}')
print(classification_report(
    y_test,
    selected_pred,
    labels=range(len(class_names)),
    target_names=class_names,
    zero_division=0
))

ConfusionMatrixDisplay.from_predictions(
    y_test,
    selected_pred,
    labels=range(len(class_names)),
    display_labels=class_names,
    xticks_rotation=45,
    cmap='Blues'
)
plt.title('Tuned SVM: Test Confusion Matrix')
plt.tight_layout()
plt.show()
```

```python
best_model_train_pred = best_model.predict(X_train)
best_model_test_pred = best_model.predict(X_test)

_, _, train_macro_f1, _ = precision_recall_fscore_support(
    y_train, best_model_train_pred, average='macro', zero_division=0
)
_, _, test_macro_f1, _ = precision_recall_fscore_support(
    y_test, best_model_test_pred, average='macro', zero_division=0
)

print(f"Train accuracy: {accuracy_score(y_train, best_model_train_pred):.3f}")
print(f"Test accuracy:  {accuracy_score(y_test, best_model_test_pred):.3f}")
print(f"Test macro F1:  {test_macro_f1:.3f}")
```
